```python
from pathlib import Path
import csv

cwd = Path.cwd()
project_root = cwd.parents[1]
input_dir = project_root / "Original_datasets" / "KiwiTrimmed"
output_dir = project_root / "Output_metadata" / "KiwiTrimmed"

output_dir.mkdir(parents=True, exist_ok=True)
out_csv = output_dir / "kiwi_metadata.csv"

rows = []

for wav in input_dir.rglob("*.wav"):
    # Subcarpeta inmediata (individual_id)
    individual_id = wav.parent.name  
    
    # Sexo a partir del nombre de la subcarpeta
    if individual_id.upper().startswith("F"):
        sex = "female"
    elif individual_id.upper().startswith("M"):
        sex = "male"
    else:
        sex = "unknown"
    
    rows.append({
        "file_name": wav.name,
        "individual_id": individual_id,
        "sex": sex
    })

# Guardar CSV
with open(out_csv, "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["file_name", "individual_id", "sex"])
    writer.writeheader()
    writer.writerows(rows)

print(f"✅ Metadata saved in: {out_csv}")
print(f"Total files: {len(rows)}")
```
